Two Pointer Approach

```python
# Q1. Move all zeros to the end
arr = [0,1, 0, 3, 0, 5, 0, 2]
write_pointer = 0
for read_pointer in range(len(arr)):
    if arr[read_pointer] != 0:
        arr[write_pointer], arr[read_pointer] = arr[read_pointer], arr[write_pointer]
        write_pointer += 1
print(arr)
```

```python
# Q2. Reverse an array

arr = [1, 2, 3, 4, 5]
i = 0 
j = len(arr)-1

while i < j:
    arr[i], arr[j] = arr[j], arr[i]
    i +=1
    j-=1
print(arr)
```

```python
# Q3. Check if an array is a palindrome

def check_palindrom(arr:list) -> bool:
    left = 0
    right = len(arr)-1
    while left < right:
        if arr[left] != arr[right]:
            return False
        left += 1
        right -= 1
    return True
    
print(check_palindrom(arr = [1, 2, 3, 2, 1]))
print(check_palindrom(arr = [1, 2, 3, 4, 1]))
```

```python
# Q4. Find a pair with a given sum Array is sorted.

def two_sum(arr:list, target:int) -> list:
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] + arr[j] == target:
                return [i, j]
    return []

def two_pointer_for_sorted(arr:list, target:int) -> list:
    left, right = 0, len(arr)-1
    while left < right:
        total = arr[left] + arr[right]
        if total == target:
            return [left, right]
        if total < target:
            left += 1
        if total > target:
            right -= 1
    return []

print(two_sum(arr = [1, 2, 4, 6, 8, 9, 11], target = 10))
print(two_pointer_for_sorted(arr = [1, 2, 4, 6, 8, 9, 11], target = 10))
```

```python
# Q5. Remove duplicates from a sorted array
def remove_duplicates(arr:list) -> list:
    slow = 0
    for fast in range(1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]
    arr [:] = arr[: slow +1 ] 
    return arr
print(remove_duplicates(arr = [1, 1, 2, 2, 3, 3, 4]))
print(remove_duplicates(arr = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4, 4]))
```

```python
# Q6. Move negative numbers to the left
def move_negative(arr:list) -> list:
    slow = 0 
    for fast in range(len(arr)):
        if arr[fast] < 0:
            # arr[slow], arr[fast] = arr[fast], arr[slow]
            arr.insert(slow, arr.pop(fast))
            slow +=1
    return arr

print(move_negative(arr = [4, -2, 7, -5, 3, -1, 8]))
```

```python
# Q7 — Pair sum, but return values
arr = [1, 2, 3, 4, 6, 8, 9]
target = 15

def two_sum(arr:list, target:int) -> list:
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
             if arr[i] + arr[j] == target:
                 return[arr[i], arr[j]]
    return []
print(two_sum(arr,target))
```

```python
# Q8 — Count pairs with a target sum
arr = [1, 2, 3, 4, 5, 6]
target = 7
def count_pair(arr:list, target:int) -> int:
    ans = 0
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
           if arr[i] + arr[j] == target:
               ans += 1
    return ans
print(count_pair(arr, target)) 
```

```python
# Q9 — Check palindrome string
def check_palindrom_string(s:str) -> bool:
    left, right = 0, len(s)-1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
print(check_palindrom_string(s = "racecar"))
print(check_palindrom_string(s = "python"))
```

```python
# Q11 — Sorted array: square and sort Given a sorted array containing negative and positive numbers:
arr = [-7, -3, -1, 4, 5, 8]
def sort_and_product(arr:list) -> list:
    inx = 0
    for i in arr:
        arr[inx] = i*i
        inx +=1
    for k in range(len(arr)):
        for j in range(k+1, len(arr)):
            if arr[k] > arr[j]:
                arr[k], arr[j] = arr[j], arr[k]
    return arr

print(sort_and_product(arr))
```
