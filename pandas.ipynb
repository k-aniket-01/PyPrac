```python
def get_weekly_range(start_date, end_date):
    start_series = pd.Series(pd.date_range(start=start_date, end=end_date, freq='W-MON', name='week_start'))
    end_series = pd.Series(pd.date_range(start=start_date, end=end_date, freq='W-SUN', name= 'week_end'))
    
    return start_series, end_series

data =  get_weekly_range(std, end)
print(data)
```
