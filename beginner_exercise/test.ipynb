```python
def fib(n):
    a = 0
    b = 1
    for _ in range(n):
        print(a, end=',')
        a,b = b,a+b
fib(7) 
```

```python
def mer_list(l1,l2):
    l1.extend(l2)
    for i in range(len(l1)):
        for j in range(len(l1)-i-1):
            if l1[j]>l1[j+1]:
                l1[j], l1[j+1] = l1[j+1],l1[j]
            print(l1)
        print()
mer_list([1,3,5],[2,4,6])
```

```python
def is_prime(n):
    prime = True
    if n == 0 or n ==1 :
        prime = False
    for i in range(2,n):
        if n%i == 0:
            prime = False
            break
        prime
    return prime
is_prime(7)
```

```python
def rotate(l,r):
    lr = l[r:]
    print(lr)
    lb = l[:r]
    print(lb)
    lr.extend(lb)
    return lr
rotate([1,2,3,4,5],2)
```

```python
def find_missing(l):
    for i in range(len(l)):
        if l[i] != i+1:
            return i+1
find_missing([1,2,3,5,6])
```

```python
def find_longest(st):
    long = ''
    for i in st.split():
        if len(i)>len(long):
            long = i
    return long

s = "I love backend development"
find_longest(s)
```

```python
def non_rep_char(c):
    d = {}
    for i in c:
        if i in d:
            d[i] += 1
        else:
            d[i] = 1
    for k,v in d.items():
        if v ==1:
            return k
c = "aabbcde"
non_rep_char(c)
```

```python
#Input "{[()]}" Output True
inp = "{[()]}"
d = {
    '{':'}',  '[':']',  '(':')'
    }
stack = []
for i in inp:
    if i in d:
        stack.append(i)
    else:
        if not stack:
            print(False)
            break
        if d[stack[-1]] == i:
            stack.pop()
        else: 
            print(False)
            break      
else:
    print(len(stack)==0)
```

```python
"""Input [1,[2,[3,4]],5]
Output [1,2,3,4,5]"""
def flatten(i):
    a = []
    for j in i:
        if isinstance(j ,list):
            a.extend(flatten(j))
        else:
            a.append(j)
    return a
n =  [1,[2,[3,4]],5]
flatten(n)
        
        
```
