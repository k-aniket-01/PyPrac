HashMap

1. Count frequency of each number.
2. Find the first duplicate.
3. Find the first non-repeating character.
4. Find two numbers whose sum equals a target.
5. Find the most frequent element.

```python
#1. Count frequency of each number.
def freq_counter_hashmap(arr:list) -> dict:
    freq = {}
    for i in arr:
        freq[i] = freq.get(i, 0) + 1
    return freq

arr = [1,2,43,56,3,321,1,123,3,4,2,4,6,5,1,23,12,23,1]
print(freq_counter_hashmap(arr))
```

```python
# 2. Find the first duplicate.
arr = [1,2,43,56,3,321,1,123,3,4,2,4,6,5,1,23,12,23,1]
def find_first_dup(arr:list) -> int:
    dict1 = {}
    for i in arr:
        if i in dict1:return i 
        else:dict1[i] = "seen"
print(find_first_dup(arr))
```

```python
#3. Find the first non-repeating character.
st = "sdkflajsdfasjfn"
d = {}
for i in st:
    d[i] = d.get(i, 0 ) + 1
for v in st:
    if d[v] == 1:
        print(v)
        break
print(d)
```

```python
#4. Find two numbers whose sum equals a target.
arr, target = [1,22,3,4,5,6,7,8,9], 5
def two_sum_hashmap(arr:list, target:int) -> list:
    d = {}
    for i,k in enumerate(arr):
        req = target - k 
        if req in d:
            return [d[req], i]
        else:
            d[k] = i
    return []
print(two_sum_hashmap(arr,target))
```

reverse a linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)
n4 = Node(40)
n5 = Node(50)
n1.next = n2 
n2.next = n3
n3.next = n4
n4.next = n5

head = n1
def rev_linked_list(head):
    start = n1
    prev = None
    while start:
        next_node = start.next
        start.next = prev
        prev = start
        start = next_node
    return prev

rev = rev_linked_list(head)    # start, start.next = start.next, prev 

    
# while rev:
#     print(rev.data)
#     rev = rev.next
    
def find_middle(head):
    slow, fast = head , head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next 
    return slow.data if slow else None
    
print(find_middle(rev))
```
