```python
# 1. Reverse a string
# Input:  "python"
# Output: "nohtyp"

def reverse_str(s:str) -> str:
    s = list(s)
    i,j = 0,len(s)-1
    while i < j:
        s[i], s[j] = s[j], s[i]
        i += 1
        j -= 1
    s = "".join(s)
    return s
print(reverse_str("python"))
```

```python
# 2. Check whether a string is a palindrome
# Input:  "madam"
# Output: True

def check_palindrom(s:str) -> bool:
    i,j = 0, len(s)-1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True
print(check_palindrom("madam"))
print(check_palindrom("dates"))
```

```python
# 3. Count vowels and consonants
# Input: "python developer"
# Output:
# Vowels: 5
# Consonants: 10

def count_vow_cons(s:str) -> dict:
    vowels = 'AEIOUaeiou'
    d = {"vowels":0, "Consonants":0}
    for i in s:
        if i in vowels:
            d["vowels"] += 1
        else:
            d["Consonants"] +=1 
    return d
print(count_vow_cons("python devloper aniket"))
```

```python
# 4. Find the largest and smallest number in a list
arr = [10, 5, 20, 8, 15]
# Don't use max() or min().

def find_min_max(arr:list) -> dict:
    largest = arr[0]
    smallest = arr[0]
    for i in arr:
        if i > largest: 
            largest = i
        if i < smallest : 
            smallest = i
    return {"largetst":largest, "smallest":smallest}
print(find_min_max(arr))
```

```python
# 9. Find whether a number is prime
# Input: 29
# Output: True
def is_prime(n:int) -> bool:
    if n <= 1:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True
print(is_prime(11))
```

```python
# 10. Generate Fibonacci numbers
# Input: 7
# Output: 0 1 1 2 3 5 8
def gen_fib(n:int) -> list:
    a,b = 0,1
    fib = []
    for i in range(n):
        fib.append(a)
        a,b = b, a+b
    return fib
print(gen_fib(7))
```

```python
# 11. Find the first non-repeating character
# Input: "aabbcdde"
# Output: "c"
def non_rep_char(s:str) -> str:
    d = {}
    for i in s:
        d[i] = d.get(i,0)+1
    for j in d:
        if d[j] == 1:
            return j
print(non_rep_char("aabbcdde"))
```

```python
# 29. Binary search Given a sorted array:
arr = [1, 3, 5, 7, 9, 11]
target = 7

def binary_serach(arr:list, target:int) -> int:
    start = 0
    end = len(arr)-1
    while start <= end:
        mid = start + (end+1) // 2
        if arr[mid] == target:
            return mid 
        if arr[mid] < target:
            end = mid
        if arr[mid] > target:
            start = mid
print(binary_serach(arr, target))
```

```python
# 33. Implement a simple LRU cache
# Support:get(key)put(key, value)
# Maximum capacity: capacity = 3

class LruCahce:
    def __init__(self):
        self.capacity = 3
        self.lru = {}
        
    def get(self, key=None):
        if key != None:
            return self.lru.get(key, None)
        return self.lru
    def put(self,key, value):
        if self.capacity > 0:
            self.lru[key] = value
            self.capacity -= 1
            return self.lru
        self.lru.pop(next(iter(self.lru)))
        self.lru[key] = value
        return self.lru
a = LruCahce()
print(a.get())
print(a.put(1,'aniket'))
print(a.put(2, "test"))
print(a.put(3, "test"))
print(a.put(4, "test"))
print(a.put(5, "test"))
        
```

```python
# 35. Group employees by department
employees = [
    {"name": "A", "dept": "IT"},    
             {"name": "B", "dept": "HR"},    
             {"name": "C", "dept": "IT"},
             {"name": "D", "dept": "HR"},]
# Output:{    "IT": ["A", "C"],    "HR": ["B", "D"]}

def group_dict_element(data:list[dict]) -> dict:
    res = {}
    for i in data:
        if i["dept"] in res:
            res[i["dept"]].append(i["name"])
        else:
            res[i["dept"]] = [i["name"]]
    return res
print(group_dict_element(employees)) 
```

```python
# 36. Implement pagination
items = list(range(1, 101))
page = 10
per_page = 5

def pagination(item:list, page:int, per_page:int) -> list:
    ost = per_page * (page - 1)
    lmt = ost + per_page 
    return item[ost:lmt]
print(pagination(items,page,per_page))
```

```python
# Find the maximum sum of a subarray of size k.

def find_max_sub(arr:list, k:int) -> list:
    start, end = 0, k
    total = 0
    res = []
    while end <= len(arr):
        s = 0
        for i in range(start, end):
            s += arr[i]
        if s > total:
            total = s
            res = arr[start:end]
        start += 1
        end += 1
    return res, total
print(find_max_sub([1,65,3,6,74,6,4,3,45,2,3,99], 3))
```

```python
arr = [[1,2,[3,4],5,7,[8,9,10],[],11]]

def flatten_list(arr:list) -> list:
    res = []
    for i in arr:
        if isinstance(i, list):
            res.extend(flatten_list(i))
        else:
            res.append(i)
    return res
print(flatten_list(arr))
```
