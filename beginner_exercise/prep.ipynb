```python
"""Input:
numbers = [2, 7, 11, 15]
target = 9
    Output:
    [0, 1]
"""
numbers = [2, 7, 11, 15]
target = 9
d ={}


for i in range(len(numbers)):
    current = numbers[i]
    needed = target - current
    if needed in d:
        print([i, d[needed]])
    else:
        d[current] = i
        

# for i in range(len(numbers)):
#     for j in range(i+1, len(numbers)):
#         if numbers[i]+numbers[j] == target:
#             print([i, j])
#             break
```

```python
a = [1,2,3,4,5]
b = [6,7,8]
m = []

x,y = 0,0

while x<len(a) and y< len(b):
    if a[x] < b[y]:
        m.append(a[x])
        x+=1
    else:
        m.append(b[y])
        y+=1
    
while x<len(a):
    m.append(a[x])
    x+=1
    
while y<len(b):
    m.append(b[y])
    y+=1
    
print(m)
```
