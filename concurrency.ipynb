```python
import multiprocessing
import os

def task(name):
    print(f"Task: {name}")
    print(f"Process ID: {os.getpid()}")

if __name__ == "__main__":
    p1 = multiprocessing.Process(
        target=task,
        args=("A",)
    )

    p2 = multiprocessing.Process(
        target=task,
        args=("B",)
    )

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("All tasks completed")
```

```python
from multiprocessing import Pool
import time


def calculate_salary(employee):
    basic = employee["basic"]

    allowances = basic * 0.20
    deductions = basic * 0.05
    tax = basic * 0.10

    net_salary = basic + allowances - deductions - tax

    return {
        "employee_id": employee["employee_id"],
        "net_salary": round(net_salary, 2)
    }


if __name__ == "__main__":

    employees = [
        {
            "employee_id": i,
            "basic": 30000 + (i % 10) * 1000
        }
        for i in range(1, 100001)
    ]

    start_time = time.perf_counter()

    with Pool(processes=10) as pool:
        payroll_results = pool.map(
            calculate_salary,
            employees
        )
    # payroll_results = []
    # for  i in employees:
    #     payroll_results.append(calculate_salary(i))

    end_time = time.perf_counter()

    print("Employees processed:", len(payroll_results))
    print("First result:", payroll_results[0])
    print("mid result:", payroll_results[len(payroll_results)//2])
    print("last result:", payroll_results[-1])
    print("Execution time:", end_time - start_time)
```

```python
from multiprocessing import Pool
import time

def calculate_salary(employee):
    basic = employee["basic"]
    # Simulate expensive CPU calculations
    score = 0
    for i in range(30_000):
        score += (i * i) % 97

    allowances = basic * 0.20
    deductions = basic * 0.05
    tax = basic * 0.10

    net_salary = basic + allowances - deductions - tax

    return {
        "employee_id": employee["employee_id"],
        "net_salary": round(net_salary, 2),
        "calculation_score": score
    }

if __name__ == "__main__":
    employees = [
        {
            "employee_id": i,
            "basic": 30000 + (i % 10) * 1000
        }
        for i in range(1, 1001)
    ]
    # Sequential execution
    start = time.perf_counter()

    sequential_results = [
        calculate_salary(employee)
        for employee in employees
    ]

    sequential_time = time.perf_counter() - start
    # Multiprocessing execution
    start = time.perf_counter()

    with Pool(processes=4) as pool:
        parallel_results = pool.map(calculate_salary, employees)

    parallel_time = time.perf_counter() - start

    print(f"Sequential time: {sequential_time:.2f}s")
    print(f"Multiprocessing time: {parallel_time:.2f}s")

    print("Results match:", sequential_results == parallel_results)

    if parallel_time > 0:
        print(
            "Speedup:",
            round(sequential_time / parallel_time, 2),
            "x"
        )
```
