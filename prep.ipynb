```python
# 1. Reverse a string
# Input:  "python"
# Output: "nohtyp"

def reverse_str(s:str) -> str:
    s = list(s)
    i,j = 0,len(s)-1
    while i < j:
        s[i], s[j] = s[j], s[i]
        i += 1
        j -= 1
    s = "".join(s)
    return s
print(reverse_str("python"))
```

```python
# 2. Check whether a string is a palindrome
# Input:  "madam"
# Output: True

def check_palindrom(s:str) -> bool:
    i,j = 0, len(s)-1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True
print(check_palindrom("madam"))
print(check_palindrom("dates"))
```

```python
# 29. Binary search Given a sorted array:
arr = [1, 3, 5, 7, 9, 11]
target = 7

def binary_serach(arr:list, target:int) -> int:
    start = 0
    end = len(arr)-1
    while start <= end:
        mid = start + (end+1) // 2
        if arr[mid] == target:
            return mid 
        if arr[mid] < target:
            end = mid
        if arr[mid] > target:
            start = mid
print(binary_serach(arr, target))
```

```python
# 33. Implement a simple LRU cache
# Support:get(key)put(key, value)
# Maximum capacity: capacity = 3

class LruCahce:
    def __init__(self):
        self.capacity = 3
        self.lru = {}
        
    def get(self, key=None):
        if key != None:
            return self.lru.get(key, None)
        return self.lru
    def put(self,key, value):
        if self.capacity > 0:
            self.lru[key] = value
            self.capacity -= 1
            return self.lru
        self.lru.pop(next(iter(self.lru)))
        self.lru[key] = value
        return self.lru
a = LruCahce()
print(a.get())
print(a.put(1,'aniket'))
print(a.put(2, "test"))
print(a.put(3, "test"))
print(a.put(4, "test"))
print(a.put(5, "test"))
        
```

```python
# Find the maximum sum of a subarray of size k.

def find_max_sub(arr:list, k:int) -> list:
    start, end = 0, k
    total = 0
    res = []
    while end <= len(arr):
        s = 0
        for i in range(start, end):
            s += arr[i]
        if s > total:
            total = s
            res = arr[start:end]
        start += 1
        end += 1
    return res, total
print(find_max_sub([1,65,3,6,74,6,4,3,45,2,3,99], 3))
```

```python
s = "A man, a plan, a canal: Panama"
# Output: true
# Explanation: "amanaplanacanalpanama" is a palindrome.

def isPalindrome( s: str) -> bool:
    clean_s = "".join(char for char in s if char.isalnum()).lower()
    i,j = 0, len(clean_s)-1
    while i < j:
        if clean_s[i] != clean_s[j]:
            return False
        i += 1
        j -= 1
    return True    
print(isPalindrome(s))
```

```python
def isValid( s: str) -> bool:
    d = {'{':'}','(':')','[':']'}
    stack = []
    for i in s:
        if i in d:
            stack.append(i)
        elif i == d[stack[-1]]:
            stack.pop()
        else:
            return False
    return len(stack) == 0

print(isValid('({[({]]]})]})'))
```

```python
numbers = [2,7,11,15]
target = 9
# Output: [1,2]
# Explanation: The sum of 2 and 7 is 9. Therefore, index1 = 1, index2 = 2. We return [1, 2].
def twoSum(numbers: list[int], target: int) -> list[int]:
    d = {}
    for i,j in enumerate(numbers, start=1):
        req = target - j
        if req in d:
            return [d[req], i]
        d[j] = i
    return []
print(twoSum(numbers,target))
```
